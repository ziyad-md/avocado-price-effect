```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

df = pd.read_csv('avocado.csv')
```

```python
# We look at 'conventional' avocados across the 'TotalUS' to remove regional and organic noise
use_data = df[(df['region'] == 'TotalUS') & (df['type'] == 'conventional')]
```

```python
sns.regplot(x = 'AveragePrice', y = 'Total Volume', data = use_data)
plt.title("Avocado Price Effect: Price vs. Volume Sold")
plt.xlabel('Average Price ($)')
plt.ylabel('Total Volume Sold (in millions)')
```

```python
# 4. Prove it mathematically with Ordinary Least Squares (OLS) regression
X = sm.add_constant(use_data['AveragePrice']) # Price is the independent variable
y = use_data['Total Volume']                  # Volume is the dependent variable

model = sm.OLS(y, X).fit()
```

```python
print(model.summary())
```
